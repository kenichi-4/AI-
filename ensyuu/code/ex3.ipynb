```python
n1 = int(input("1 つ目の数値を入力してください："))
n2 = int(input("2 つ目の数値を入力してください："))
n3 = int(input("3 つ目の数値を入力してください："))
numbers = [n1,n2,n3]
print(max(numbers))
print(min(numbers))
print(sum(numbers)/len(numbers))
```

```python
a1 = int(input("グループ A の 1 つ目の数値を入力してください："))
a2 = int(input("グループ A の 2 つ目の数値を入力してください："))
b1 = int(input("グループ B の 1 つ目の数値を入力してください："))
b2 = int(input("グループ B の 2 つ目の数値を入力してください："))
a = {a1,a2}
b = {b1,b2}
print("どちらかのグループに含まれる値は以下の通りです。")
print(a|b)
print("どちらのグループにも含まれる値は以下の通りです。")
print(a&b)
```

```python
n = int(input("4 桁の数値を入力してください"))
one_place = n%10
ten_place = n%100 // 10
hun_place = n%1000 // 100
thou_place = n // 1000
print("1 の位は:" + str(one_place))
print("10 の位は:" + str(ten_place))
print("100 の位は:" + str(hun_place))
print("1000 の位は:" + str(thou_place))
```
